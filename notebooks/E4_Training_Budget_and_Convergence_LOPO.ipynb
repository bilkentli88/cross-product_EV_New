# E4 — Training-budget and convergence sensitivity
## Does the topology-mechanism pattern survive substantially longer training?

The primary manuscript and E1--E3 use a fixed 30-epoch training budget, corresponding to 270 product-level optimizer updates per neural run. E4 tests whether the central mechanism pattern is an artifact of undertraining, differential convergence speed, or a regularization effect that disappears with longer optimization.

### Conditions

Four key graph conditions are followed along the same training trajectory:

1. **RealGNN** — recorded physical fixation topology.
2. **NoEdgesGNN** — no cross-node edges.
3. **CompleteGraphGNN** — all-to-all cross-node interaction with a dummy edge type and no physical locality.
4. **ShuffledGNN × 5** — the same five semantic--topology permutations used in the primary study.

### Training budgets

Each trajectory is evaluated after:

- **30 epochs** — the manuscript's original budget;
- **100 epochs**;
- **300 epochs**.

The model is **not restarted** between budgets: the same deterministic training trajectory is continued from epoch 1 through epoch 300. This directly measures what happens when the available optimization budget is increased.

### Fixed optimization settings

To isolate training budget, the following remain unchanged:

- AdamW;
- learning rate `2e-3`;
- weight decay `1e-4`;
- dropout `0.10`;
- eight sampled negatives per training state;
- the same 10 LOPO folds;
- neural-training seeds `[17, 29, 43]`;
- the same five topology-permutation seeds.

No held-out-product information is used for model selection or stopping.

### Outputs

E4 records:

- Hits@1, Hits@3, and MRR at epochs 30, 100, and 300;
- epoch-level training loss for every trajectory;
- product-level exact paired contrasts at each budget;
- product-level changes from 30→100, 100→300, and 30→300 epochs.

The held-out battery remains the independent unit (`n=10`). Seeds and permutation realizations are averaged within product before inference.

This is a robustness/mechanism diagnostic. Its purpose is not to select the best epoch retrospectively, but to determine whether the qualitative mechanism conclusion is stable over a much larger optimization budget.


```python
# 1. Imports and frozen configuration

from pathlib import Path
from collections import Counter, defaultdict
import tarfile, json, math, random, time

import numpy as np
import pandas as pd

import torch
import torch.nn as nn
import torch.nn.functional as F

SEEDS = [17, 29, 43]
SHUFFLE_IDS = [1, 2, 3, 4, 5]

EPOCHS = 30
NEGATIVES_PER_STATE = 8
LR = 2e-3
WEIGHT_DECAY = 1e-4
DROPOUT = 0.10

EMBED_DIM = 12
HIDDEN_DIM = 64
GNN_DIM = 32

BASE_SEED = 20261003
GRAPH_SHUFFLE_BASE = 20261119
REWIRE_BASE = 20270117  # unused in E4; retained for helper compatibility


DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

EVAL_EPOCHS = [30, 100, 300]
MAX_EPOCHS = max(EVAL_EPOCHS)

# One trajectory each for real, no-edge, complete, and five permutations.
N_TRAJECTORIES = 10 * len(SEEDS) * (3 + len(SHUFFLE_IDS))

print("Device:", DEVICE)
print("Torch:", torch.__version__)
print("Planned E4 training trajectories:", N_TRAJECTORIES)
print("Evaluation checkpoints:", EVAL_EPOCHS)
print("Maximum epochs per trajectory:", MAX_EPOCHS)

if DEVICE.type != "cuda":
    print("WARNING: GPU is strongly recommended for this notebook.")
```

```python
# 3. Load and validate raw data.

bat = pd.read_csv(DATA / "battery_catalogue.csv")
parts = pd.read_csv(DATA / "parts.csv")
fix = pd.read_csv(DATA / "fixations.csv")
ops = pd.read_csv(DATA / "operations.csv")

print("Batteries:", len(bat))
print("Parts:", len(parts))
print("Fixations:", len(fix))
print("Operations:", len(ops))

display(bat[["id", "name"]])
```

```python
# 4. Attach each operation to a battery and reconstruct the 10 reference chains.

battery_names = bat.set_index("id")["name"].to_dict()

fix_battery = fix.set_index("id")["battery_id"].to_dict()
part_battery = parts.set_index("id")["battery_id"].to_dict()
part_class_raw = parts.set_index("id")["class"].to_dict()

fix_info_raw = (
    fix.set_index("id")[["class", "fromPartID", "toPartID"]]
    .to_dict("index")
)

def operation_battery(row):
    if pd.notna(row.fixation_id):
        return int(fix_battery[int(row.fixation_id)])
    return int(part_battery[int(row.part_id)])

ops["battery_id"] = ops.apply(operation_battery, axis=1)
ops_by_id = ops.set_index("id")

chains = {}

for b in sorted(ops["battery_id"].unique()):
    sub = ops[ops["battery_id"] == b]
    ids = set(sub["id"].astype(int))
    next_ids = set(sub["next_op_id"].dropna().astype(int))

    starts = list(ids - next_ids)
    terminals = sub[sub["next_op_id"].isna()]["id"].astype(int).tolist()

    assert len(starts) == 1
    assert len(terminals) == 1

    seq = []
    seen = set()
    cur = int(starts[0])

    while cur in ids and cur not in seen:
        seen.add(cur)
        seq.append(cur)

        nxt = ops_by_id.loc[cur, "next_op_id"]
        if pd.isna(nxt):
            break
        cur = int(nxt)

    assert len(seq) == len(ids), f"Broken chain for battery {b}"
    chains[int(b)] = seq

print({
    battery_names[b]: len(seq)
    for b, seq in chains.items()
})
```

```python
# 5. Raw full relational signatures.
#
# These are used only for diagnostics (seen vs unseen motif/signature).
# They are NOT output classes for the neural ranking models.

def raw_full_signature(op_id):
    r = ops_by_id.loc[int(op_id)]
    skill = str(r.skill_id)

    if pd.notna(r.fixation_id):
        fi = fix_info_raw[int(r.fixation_id)]
        src_cls = str(part_class_raw[int(fi["fromPartID"])])
        dst_cls = str(part_class_raw[int(fi["toPartID"])])
        return (
            f"{skill}|fix:{fi['class']}|"
            f"{src_cls}->{dst_cls}"
        )

    return (
        f"{skill}|part:"
        f"{part_class_raw[int(r.part_id)]}"
    )

raw_signatures = {
    b: [raw_full_signature(op_id) for op_id in seq]
    for b, seq in chains.items()
}
```

```python
# 6. Fold-specific vocabularies.
#
# Vocabularies are learned ONLY from the 9 training batteries.
# Category 0 is <UNK>, so truly unseen test categories do not leak into training.

def make_vocab(values):
    vals = sorted({
        str(x)
        for x in values
        if pd.notna(x)
    })
    return {
        "<UNK>": 0,
        **{v: i + 1 for i, v in enumerate(vals)}
    }

def lookup(vocab, value):
    if pd.isna(value):
        return 0
    return vocab.get(str(value), 0)

def build_fold_vocab(train_batteries):
    return {
        "part": make_vocab(
            parts[
                parts["battery_id"].isin(train_batteries)
            ]["class"]
        ),
        "fix": make_vocab(
            fix[
                fix["battery_id"].isin(train_batteries)
            ]["class"]
        ),
        "skill": make_vocab(
            ops[
                ops["battery_id"].isin(train_batteries)
            ]["skill_id"]
        ),
    }
```

```python
# 7. Convert one battery into product-invariant graph + operation features.
#
# graph_mode:
#   "real"    -> true physical fixation topology
#   "shuffle" -> fixed permutation of topology relative to semantic/operation nodes
#   "noedge"  -> same GNN architecture, but no message-passing edges
#
# Operation-local attributes are identical across all graph modes.
# Therefore only the graph branch is manipulated.

def build_battery_pack(
    battery_id,
    vocab,
    graph_mode="real",
    shuffle_id=0,
):
    assert graph_mode in {"real", "shuffle", "noedge"}

    psub = parts[
        parts["battery_id"] == battery_id
    ].copy()

    fsub = fix[
        fix["battery_id"] == battery_id
    ].copy()

    part_id_to_local = {
        int(pid): i
        for i, pid in enumerate(psub["id"].astype(int))
    }

    n_nodes = len(psub)

    true_degree = np.zeros(n_nodes, dtype=np.float32)

    edge_src = []
    edge_dst = []
    edge_type = []

    for _, r in fsub.iterrows():
        a = part_id_to_local[int(r.fromPartID)]
        b = part_id_to_local[int(r.toPartID)]

        true_degree[a] += 1
        true_degree[b] += 1

        # Undirected message-passing representation.
        edge_src += [a, b]
        edge_dst += [b, a]

        et = lookup(vocab["fix"], r["class"])
        edge_type += [et, et]

    true_degree_norm = (
        true_degree / max(1.0, float(true_degree.max()))
    )

    node_part_class = np.array(
        [
            lookup(vocab["part"], cls)
            for cls in psub["class"]
        ],
        dtype=np.int64,
    )

    true_edge_index = np.array(
        [edge_src, edge_dst],
        dtype=np.int64,
    )

    true_edge_type = np.array(
        edge_type,
        dtype=np.int64,
    )

    if graph_mode == "real":
        graph_edge_index = true_edge_index.copy()
        graph_edge_type = true_edge_type.copy()
        graph_degree_norm = true_degree_norm.copy()

    elif graph_mode == "shuffle":
        # A deterministic, independent permutation for each
        # (shuffle_id, battery). This preserves the unlabeled graph,
        # but reassigns its physical locations to semantic nodes.
        rng_graph = np.random.default_rng(
            GRAPH_SHUFFLE_BASE
            + 10007 * int(shuffle_id)
            + int(battery_id)
        )

        permutation = rng_graph.permutation(n_nodes)

        graph_edge_index = permutation[true_edge_index]
        graph_edge_type = true_edge_type.copy()

        shuffled_degree = np.bincount(
            graph_edge_index.reshape(-1),
            minlength=n_nodes,
        ).astype(np.float32)

        graph_degree_norm = (
            shuffled_degree
            / max(1.0, float(shuffled_degree.max()))
        )

    else:  # noedge
        graph_edge_index = np.empty((2, 0), dtype=np.int64)
        graph_edge_type = np.empty((0,), dtype=np.int64)

        # Do not leak adjacency through the graph branch.
        graph_degree_norm = np.zeros(
            n_nodes,
            dtype=np.float32,
        )

    seq = [int(x) for x in chains[battery_id]]

    categorical = []
    graph_nodes = []
    numeric = []
    signatures = []

    for op_id in seq:
        r = ops_by_id.loc[op_id]

        skill_idx = lookup(
            vocab["skill"],
            r.skill_id,
        )

        if pd.notna(r.fixation_id):
            fi = fix_info_raw[int(r.fixation_id)]

            src_pid = int(fi["fromPartID"])
            dst_pid = int(fi["toPartID"])

            src_node = part_id_to_local[src_pid]
            dst_node = part_id_to_local[dst_pid]

            categorical.append([
                skill_idx,
                0,
                lookup(vocab["fix"], fi["class"]),
                lookup(vocab["part"], part_class_raw[src_pid]),
                lookup(vocab["part"], part_class_raw[dst_pid]),
                0,
            ])

            graph_nodes.append([
                src_node,
                dst_node,
                -1,
            ])

            # IMPORTANT:
            # identical operation-local features in every graph condition.
            numeric.append([
                true_degree_norm[src_node],
                true_degree_norm[dst_node],
                0.0,
            ])

        else:
            pid = int(r.part_id)
            node = part_id_to_local[pid]

            categorical.append([
                skill_idx,
                1,
                0,
                0,
                0,
                lookup(vocab["part"], part_class_raw[pid]),
            ])

            graph_nodes.append([
                -1,
                -1,
                node,
            ])

            numeric.append([
                0.0,
                0.0,
                true_degree_norm[node],
            ])

        signatures.append(raw_full_signature(op_id))

    return {
        "battery_id": int(battery_id),
        "battery_name": battery_names[int(battery_id)],
        "node_part_class": node_part_class,
        "node_degree": graph_degree_norm.astype(np.float32),
        "edge_index": graph_edge_index,
        "edge_type": graph_edge_type,
        "op_categorical": np.asarray(categorical, dtype=np.int64),
        "op_graph_nodes": np.asarray(graph_nodes, dtype=np.int64),
        "op_numeric": np.asarray(numeric, dtype=np.float32),
        "full_signatures": signatures,
        "n_ops": len(seq),
        "graph_mode": graph_mode,
        "shuffle_id": int(shuffle_id),
    }
```

```python
# 8. Neural models.

class BaseOperationRanker(nn.Module):
    """
    Product-invariant MLP operation ranker.

    It never receives battery/product IDs or raw entity IDs.
    """

    def __init__(
        self,
        n_skill,
        n_fix,
        n_part,
        emb_dim=EMBED_DIM,
        hidden_dim=HIDDEN_DIM,
        dropout=DROPOUT,
    ):
        super().__init__()

        self.skill_emb = nn.Embedding(n_skill, emb_dim)
        self.kind_emb = nn.Embedding(2, 4)
        self.fix_emb = nn.Embedding(n_fix, emb_dim)
        self.part_emb = nn.Embedding(n_part, emb_dim)

        op_input_dim = (
            emb_dim      # skill
            + 4          # operation kind
            + emb_dim    # fixation class
            + emb_dim    # source part class
            + emb_dim    # target part class
            + emb_dim    # removed part class
            + 3          # normalized local degrees
        )

        self.op_mlp = nn.Sequential(
            nn.Linear(op_input_dim, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
        )

        # current, candidate, abs difference, product, progress
        self.score_mlp = nn.Sequential(
            nn.Linear(hidden_dim * 4 + 1, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, 1),
        )

    def base_operation_encoding(self, pack):
        cats = torch.as_tensor(
            pack["op_categorical"],
            dtype=torch.long,
            device=DEVICE,
        )

        nums = torch.as_tensor(
            pack["op_numeric"],
            dtype=torch.float32,
            device=DEVICE,
        )

        skill = cats[:, 0]
        kind = cats[:, 1]
        fix_class = cats[:, 2]
        src_class = cats[:, 3]
        dst_class = cats[:, 4]
        part_class = cats[:, 5]

        x = torch.cat([
            self.skill_emb(skill),
            self.kind_emb(kind),
            self.fix_emb(fix_class),
            self.part_emb(src_class),
            self.part_emb(dst_class),
            self.part_emb(part_class),
            nums,
        ], dim=-1)

        return self.op_mlp(x)

    def encode_operations(self, pack):
        return self.base_operation_encoding(pack)

    def score_pairs(self, current_repr, candidate_repr, progress):
        x = torch.cat([
            current_repr,
            candidate_repr,
            torch.abs(current_repr - candidate_repr),
            current_repr * candidate_repr,
            progress,
        ], dim=-1)

        return self.score_mlp(x).squeeze(-1)


class GraphEncoder(nn.Module):
    """
    Two-layer product-invariant message-passing network.
    """

    def __init__(
        self,
        n_part,
        n_fix,
        part_emb_dim=EMBED_DIM,
        edge_emb_dim=8,
        graph_dim=GNN_DIM,
        dropout=DROPOUT,
    ):
        super().__init__()

        self.part_emb = nn.Embedding(n_part, part_emb_dim)
        self.edge_emb = nn.Embedding(n_fix, edge_emb_dim)

        self.input_proj = nn.Linear(
            part_emb_dim + 1,
            graph_dim,
        )

        self.message_layers = nn.ModuleList([
            nn.Linear(
                graph_dim + edge_emb_dim,
                graph_dim,
            )
            for _ in range(2)
        ])

        self.self_layers = nn.ModuleList([
            nn.Linear(graph_dim, graph_dim)
            for _ in range(2)
        ])

        self.dropout = dropout

    def forward(self, pack):
        node_class = torch.as_tensor(
            pack["node_part_class"],
            dtype=torch.long,
            device=DEVICE,
        )

        degree = torch.as_tensor(
            pack["node_degree"],
            dtype=torch.float32,
            device=DEVICE,
        ).unsqueeze(-1)

        h = F.relu(
            self.input_proj(
                torch.cat([
                    self.part_emb(node_class),
                    degree,
                ], dim=-1)
            )
        )

        edge_index = torch.as_tensor(
            pack["edge_index"],
            dtype=torch.long,
            device=DEVICE,
        )

        edge_type = torch.as_tensor(
            pack["edge_type"],
            dtype=torch.long,
            device=DEVICE,
        )

        src = edge_index[0]
        dst = edge_index[1]

        edge_embedding = self.edge_emb(edge_type)

        for message_layer, self_layer in zip(
            self.message_layers,
            self.self_layers,
        ):
            msg = message_layer(
                torch.cat([
                    h[src],
                    edge_embedding,
                ], dim=-1)
            )

            agg = torch.zeros(
                (h.shape[0], h.shape[1]),
                dtype=h.dtype,
                device=DEVICE,
            )

            agg.index_add_(0, dst, msg)

            counts = torch.zeros(
                (h.shape[0], 1),
                dtype=h.dtype,
                device=DEVICE,
            )

            counts.index_add_(
                0,
                dst,
                torch.ones(
                    (dst.numel(), 1),
                    dtype=h.dtype,
                    device=DEVICE,
                ),
            )

            agg = agg / counts.clamp_min(1.0)

            h = F.relu(
                self_layer(h) + agg
            )

            h = F.dropout(
                h,
                p=self.dropout,
                training=self.training,
            )

        return h


class GNNOperationRanker(BaseOperationRanker):

    def __init__(
        self,
        n_skill,
        n_fix,
        n_part,
        emb_dim=EMBED_DIM,
        hidden_dim=HIDDEN_DIM,
        graph_dim=GNN_DIM,
        dropout=DROPOUT,
    ):
        super().__init__(
            n_skill=n_skill,
            n_fix=n_fix,
            n_part=n_part,
            emb_dim=emb_dim,
            hidden_dim=hidden_dim,
            dropout=dropout,
        )

        self.graph_encoder = GraphEncoder(
            n_part=n_part,
            n_fix=n_fix,
            part_emb_dim=emb_dim,
            graph_dim=graph_dim,
            dropout=dropout,
        )

        # src embedding, dst embedding, absolute difference
        self.structural_projection = nn.Sequential(
            nn.Linear(graph_dim * 3, hidden_dim),
            nn.ReLU(),
        )

        self.combine = nn.Sequential(
            nn.Linear(hidden_dim * 2, hidden_dim),
            nn.ReLU(),
        )

    def encode_operations(self, pack):
        base = self.base_operation_encoding(pack)
        node_h = self.graph_encoder(pack)

        op_nodes = torch.as_tensor(
            pack["op_graph_nodes"],
            dtype=torch.long,
            device=DEVICE,
        )

        n_ops = op_nodes.shape[0]

        a = torch.zeros(
            (n_ops, node_h.shape[1]),
            device=DEVICE,
        )

        b = torch.zeros_like(a)

        fixation_mask = op_nodes[:, 0] >= 0

        if fixation_mask.any():
            a[fixation_mask] = node_h[
                op_nodes[fixation_mask, 0]
            ]

            b[fixation_mask] = node_h[
                op_nodes[fixation_mask, 1]
            ]

        part_mask = op_nodes[:, 2] >= 0

        if part_mask.any():
            a[part_mask] = node_h[
                op_nodes[part_mask, 2]
            ]

        structural = self.structural_projection(
            torch.cat([
                a,
                b,
                torch.abs(a - b),
            ], dim=-1)
        )

        return self.combine(
            torch.cat([
                base,
                structural,
            ], dim=-1)
        )
```

```python
# 9. Reproducibility + pairwise-ranking training

def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


def make_training_pairs(n_ops, k, rng):
    current, positive, negative, progress = [], [], [], []

    for t in range(n_ops - 2):
        pool = np.arange(
            t + 2,
            n_ops,
            dtype=np.int64,
        )

        neg = rng.choice(
            pool,
            size=k,
            replace=(len(pool) < k),
        )

        current.append(t)
        positive.append(t + 1)
        negative.append(neg)
        progress.append(
            t / max(1, n_ops - 1)
        )

    return (
        np.asarray(current, dtype=np.int64),
        np.asarray(positive, dtype=np.int64),
        np.stack(negative),
        np.asarray(progress, dtype=np.float32),
    )


def build_gnn(vocab):
    model = GNNOperationRanker(
        n_skill=len(vocab["skill"]),
        n_fix=len(vocab["fix"]),
        n_part=len(vocab["part"]),
    )
    return model.to(DEVICE)


def train_one_gnn_condition(
    train_batteries,
    test_battery,
    seed,
    graph_mode,
    shuffle_id=0,
):
    set_seed(seed)

    vocab = build_fold_vocab(train_batteries)

    packs = {
        b: build_battery_pack(
            b,
            vocab,
            graph_mode=graph_mode,
            shuffle_id=shuffle_id,
        )
        for b in train_batteries + [test_battery]
    }

    model = build_gnn(vocab)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY,
    )

    # Using the same training seed across paired graph conditions
    # keeps initialization, training order, and negative sampling paired.
    rng = np.random.default_rng(seed)

    for epoch in range(EPOCHS):
        order = list(train_batteries)
        rng.shuffle(order)

        model.train()

        for b in order:
            pack = packs[b]

            cur, pos, neg, prog = make_training_pairs(
                pack["n_ops"],
                NEGATIVES_PER_STATE,
                rng,
            )

            z = model.encode_operations(pack)

            cur_t = torch.as_tensor(
                cur,
                dtype=torch.long,
                device=DEVICE,
            )

            pos_t = torch.as_tensor(
                pos,
                dtype=torch.long,
                device=DEVICE,
            )

            neg_t = torch.as_tensor(
                neg,
                dtype=torch.long,
                device=DEVICE,
            )

            prog_t = torch.as_tensor(
                prog,
                dtype=torch.float32,
                device=DEVICE,
            ).unsqueeze(-1)

            cur_z = z[cur_t]
            pos_z = z[pos_t]

            pos_score = model.score_pairs(
                cur_z,
                pos_z,
                prog_t,
            )

            k = neg_t.shape[1]

            cur_rep = (
                cur_z[:, None, :]
                .expand(-1, k, -1)
                .reshape(-1, cur_z.shape[1])
            )

            neg_z = z[neg_t.reshape(-1)]

            prog_rep = (
                prog_t[:, None, :]
                .expand(-1, k, -1)
                .reshape(-1, 1)
            )

            neg_score = model.score_pairs(
                cur_rep,
                neg_z,
                prog_rep,
            ).view(len(cur), k)

            loss = F.softplus(
                -(pos_score[:, None] - neg_score)
            ).mean()

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

    return model, packs[test_battery], vocab
```

```python
# 10. Tie-aware ranking metrics.

def expected_tie_metrics(scores, target_position=0, tol=1e-8):
    """
    Scores correspond to candidate operations.

    Returns the expected Hits@1, Hits@3, and reciprocal rank
    under random ordering within exact/near score ties.
    """

    scores = np.asarray(scores)
    target_score = scores[target_position]

    greater = int(
        np.sum(scores > target_score + tol)
    )

    equal = int(
        np.sum(
            np.abs(scores - target_score) <= tol
        )
    )

    equal = max(equal, 1)

    hits1 = (
        max(
            0,
            min(equal, 1 - greater),
        )
        / equal
        if greater < 1
        else 0.0
    )

    hits3 = (
        max(
            0,
            min(equal, 3 - greater),
        )
        / equal
        if greater < 3
        else 0.0
    )

    ranks = np.arange(
        greater + 1,
        greater + equal + 1,
        dtype=float,
    )

    mrr = float(
        np.mean(1.0 / ranks)
    )

    return hits1, hits3, mrr
```

```python
# 11. Evaluate one neural model on an unseen battery.

def training_signature_set(train_batteries):
    s = set()

    for b in train_batteries:
        s.update(raw_signatures[b])

    return s


def evaluate_neural_model(
    model,
    test_pack,
    train_batteries,
):
    model.eval()

    seen_signatures = training_signature_set(
        train_batteries
    )

    rows = []

    with torch.no_grad():
        operation_repr = model.encode_operations(
            test_pack
        )

        n_ops = test_pack["n_ops"]

        for t in range(n_ops - 1):
            candidate_indices = np.arange(
                t + 1,
                n_ops,
                dtype=np.int64,
            )

            candidate_t = torch.as_tensor(
                candidate_indices,
                dtype=torch.long,
                device=DEVICE,
            )

            current_repr = (
                operation_repr[t]
                .unsqueeze(0)
                .expand(
                    len(candidate_indices),
                    -1,
                )
            )

            candidate_repr = operation_repr[
                candidate_t
            ]

            progress = torch.full(
                (len(candidate_indices), 1),
                t / max(1, n_ops - 1),
                dtype=torch.float32,
                device=DEVICE,
            )

            scores = model.score_pairs(
                current_repr,
                candidate_repr,
                progress,
            ).cpu().numpy()

            # candidate_indices[0] is the reference next operation.
            h1, h3, mrr = expected_tie_metrics(
                scores,
                target_position=0,
            )

            target_signature = (
                test_pack["full_signatures"][t + 1]
            )

            rows.append({
                "position": t,
                "hits1": h1,
                "hits3": h3,
                "mrr": mrr,
                "signature_seen_in_training":
                    target_signature in seen_signatures,
            })

    detail = pd.DataFrame(rows)

    metrics = {
        "hits1": float(detail["hits1"].mean()),
        "hits3": float(detail["hits3"].mean()),
        "mrr": float(detail["mrr"].mean()),
        "seen_signature_rate": float(
            detail["signature_seen_in_training"].mean()
        ),
    }

    seen = detail[
        detail["signature_seen_in_training"]
    ]

    unseen = detail[
        ~detail["signature_seen_in_training"]
    ]

    for label, subset in [
        ("seen", seen),
        ("unseen", unseen),
    ]:
        if len(subset) > 0:
            metrics[f"{label}_hits1"] = float(
                subset["hits1"].mean()
            )

            metrics[f"{label}_hits3"] = float(
                subset["hits3"].mean()
            )

            metrics[f"{label}_mrr"] = float(
                subset["mrr"].mean()
            )
        else:
            metrics[f"{label}_hits1"] = np.nan
            metrics[f"{label}_hits3"] = np.nan
            metrics[f"{label}_mrr"] = np.nan

    return metrics, detail
```

```python
# 12. Exact-capacity validity helper

def count_trainable_parameters(model):
    return sum(
        p.numel()
        for p in model.parameters()
        if p.requires_grad
    )
```

## 13. E2 graph builders

`build_battery_pack` remains unchanged for the canonical real, shuffled, and no-edge conditions.

The helper below creates the three new E2 interventions:

- `meanpool`: no cross-node edges and zero graph-branch degree;
- `complete`: directed all-to-all edges excluding self-loops, one dummy edge type, constant graph-branch degree;
- `rewire`: component-class-preserving endpoint swaps on fixation records.

Operation-level semantic inputs and operation-local normalized physical-degree features remain exactly the same as in the primary experiment.


```python
# 13. New E2 graph-pack builders

from collections import defaultdict

def _part_local_maps(battery_id):
    psub = parts[parts["battery_id"] == battery_id].copy()
    part_ids = psub["id"].astype(int).tolist()
    part_id_to_local = {pid: i for i, pid in enumerate(part_ids)}
    local_raw_class = {
        i: str(cls)
        for i, cls in enumerate(psub["class"].tolist())
    }
    return psub, part_id_to_local, local_raw_class


def _connected_components_count(n_nodes, undirected_edges):
    adj = [[] for _ in range(n_nodes)]
    for u, v in undirected_edges:
        if u == v:
            continue
        adj[u].append(v)
        adj[v].append(u)

    seen = set()
    n_comp = 0
    for start in range(n_nodes):
        if start in seen:
            continue
        n_comp += 1
        stack = [start]
        seen.add(start)
        while stack:
            u = stack.pop()
            for v in adj[u]:
                if v not in seen:
                    seen.add(v)
                    stack.append(v)
    return n_comp


def class_preserving_rewire_records(
    battery_id,
    vocab,
    rewire_id,
    swaps_per_edge=20,
):
    """
    Degree-preserving endpoint swaps on fixation records.

    We swap two selected endpoints only when their raw component classes are
    identical. Edge type stays attached to each fixation record.

    Preserved exactly:
      - node fixation degree,
      - number of fixation records,
      - edge-type multiset,
      - ordered endpoint-class pair of each fixation record.

    Parallel edges are allowed because the source graph is a multigraph.
    Self-loops are rejected.
    """
    psub, part_id_to_local, local_raw_class = _part_local_maps(battery_id)
    fsub = fix[fix["battery_id"] == battery_id].copy()

    records = []
    for _, r in fsub.iterrows():
        u = part_id_to_local[int(r.fromPartID)]
        v = part_id_to_local[int(r.toPartID)]
        et = lookup(vocab["fix"], r["class"])
        records.append([u, v, et])

    original = [x.copy() for x in records]
    n_nodes = len(psub)
    n_edges = len(records)

    original_degree = np.zeros(n_nodes, dtype=np.int64)
    for u, v, _ in records:
        original_degree[u] += 1
        original_degree[v] += 1

    original_components = _connected_components_count(
        n_nodes,
        [(u, v) for u, v, _ in records],
    )

    rng = np.random.default_rng(
        REWIRE_BASE + 10007 * int(rewire_id) + int(battery_id)
    )

    endpoint_refs = defaultdict(list)
    for ei, (u, v, _) in enumerate(records):
        endpoint_refs[local_raw_class[u]].append((ei, 0))
        endpoint_refs[local_raw_class[v]].append((ei, 1))

    eligible_classes = [
        cls for cls, refs in endpoint_refs.items()
        if len(refs) >= 2
    ]
    assert eligible_classes, f"No eligible class-preserving swaps for battery {battery_id}"

    target = max(1, swaps_per_edge * n_edges)
    max_attempts = max(target * 50, 10000)
    accepted = 0
    attempts = 0

    while accepted < target and attempts < max_attempts:
        attempts += 1
        cls = eligible_classes[int(rng.integers(0, len(eligible_classes)))]
        refs = endpoint_refs[cls]

        idx = rng.choice(len(refs), size=2, replace=False)
        (e1, p1), (e2, p2) = refs[int(idx[0])], refs[int(idx[1])]

        if e1 == e2:
            continue

        n1 = records[e1][p1]
        n2 = records[e2][p2]

        if n1 == n2:
            continue

        other1 = records[e1][1 - p1]
        other2 = records[e2][1 - p2]

        # Reject self-loops after swapping.
        if n2 == other1 or n1 == other2:
            continue

        # Because n1 and n2 have the same raw class, each edge record keeps
        # its original ordered endpoint-class pair.
        records[e1][p1] = n2
        records[e2][p2] = n1
        accepted += 1

    new_degree = np.zeros(n_nodes, dtype=np.int64)
    for u, v, _ in records:
        new_degree[u] += 1
        new_degree[v] += 1

    assert np.array_equal(original_degree, new_degree)

    # Every fixation record must preserve its endpoint classes and edge type.
    for old, new in zip(original, records):
        ou, ov, ot = old
        nu, nv, nt = new
        assert ot == nt
        assert local_raw_class[ou] == local_raw_class[nu]
        assert local_raw_class[ov] == local_raw_class[nv]

    changed_records = sum(
        1 for old, new in zip(original, records)
        if old[0] != new[0] or old[1] != new[1]
    )
    changed_fraction = changed_records / max(1, n_edges)

    new_components = _connected_components_count(
        n_nodes,
        [(u, v) for u, v, _ in records],
    )

    return {
        "records": records,
        "accepted_swaps": int(accepted),
        "attempts": int(attempts),
        "changed_fraction": float(changed_fraction),
        "original_components": int(original_components),
        "rewired_components": int(new_components),
    }


def build_e2_pack(
    battery_id,
    vocab,
    condition,
    realization_id=0,
):
    """
    Build one battery pack for any E2 condition.
    """
    if condition == "real":
        return build_battery_pack(
            battery_id, vocab, graph_mode="real", shuffle_id=0
        )

    if condition == "noedge":
        return build_battery_pack(
            battery_id, vocab, graph_mode="noedge", shuffle_id=0
        )

    if condition == "shuffle":
        return build_battery_pack(
            battery_id, vocab, graph_mode="shuffle", shuffle_id=realization_id
        )

    # Start from the real pack so operation inputs remain byte-for-byte
    # identical to the primary experiment.
    pack = build_battery_pack(
        battery_id, vocab, graph_mode="real", shuffle_id=0
    )

    n_nodes = len(pack["node_part_class"])

    if condition == "meanpool":
        pack["edge_index"] = np.empty((2, 0), dtype=np.int64)
        pack["edge_type"] = np.empty((0,), dtype=np.int64)
        pack["node_degree"] = np.zeros(n_nodes, dtype=np.float32)
        pack["graph_mode"] = "meanpool"
        pack["realization_id"] = 0
        return pack

    if condition == "complete":
        src = []
        dst = []
        for u in range(n_nodes):
            for v in range(n_nodes):
                if u != v:
                    src.append(u)
                    dst.append(v)

        pack["edge_index"] = np.asarray([src, dst], dtype=np.int64)
        # Dummy/unknown edge type only; no physical fixation type information.
        pack["edge_type"] = np.zeros(len(src), dtype=np.int64)
        # Complete-graph normalized degree is constant for every node.
        pack["node_degree"] = np.ones(n_nodes, dtype=np.float32)
        pack["graph_mode"] = "complete"
        pack["realization_id"] = 0
        return pack

    if condition == "rewire":
        info = class_preserving_rewire_records(
            battery_id=battery_id,
            vocab=vocab,
            rewire_id=realization_id,
        )

        edge_src = []
        edge_dst = []
        edge_type = []

        for u, v, et in info["records"]:
            edge_src += [u, v]
            edge_dst += [v, u]
            edge_type += [et, et]

        pack["edge_index"] = np.asarray(
            [edge_src, edge_dst],
            dtype=np.int64,
        )
        pack["edge_type"] = np.asarray(edge_type, dtype=np.int64)

        # Degree sequence is preserved exactly, so the real-pack degree input
        # remains valid and identical node by node.
        pack["graph_mode"] = "rewire"
        pack["realization_id"] = int(realization_id)
        pack["rewire_accepted_swaps"] = info["accepted_swaps"]
        pack["rewire_attempts"] = info["attempts"]
        pack["rewire_changed_fraction"] = info["changed_fraction"]
        pack["rewire_original_components"] = info["original_components"]
        pack["rewire_components"] = info["rewired_components"]
        return pack

    raise ValueError(f"Unknown E2 condition: {condition}")
```

## 15. E4 model builder and evaluation-safe training loop

The training loop below mirrors the canonical notebook, but continues each model to 300 epochs and evaluates at fixed checkpoints. Evaluation uses `model.eval()` and does not alter the negative-sampling RNG, so the training trajectory remains paired across conditions.


```python
# 15. E4 model builder and one-trajectory trainer

import itertools
import shutil

def build_e4_model(vocab):
    return GNNOperationRanker(
        n_skill=len(vocab["skill"]),
        n_fix=len(vocab["fix"]),
        n_part=len(vocab["part"]),
    ).to(DEVICE)


def train_e4_trajectory(
    train_batteries,
    test_battery,
    seed,
    condition,
    realization_id=0,
):
    """
    Train one graph condition from epoch 1 to MAX_EPOCHS and evaluate
    at every checkpoint in EVAL_EPOCHS.

    condition ∈ {"real", "noedge", "complete", "shuffle"}.
    """
    assert condition in {"real", "noedge", "complete", "shuffle"}

    set_seed(seed)

    vocab = build_fold_vocab(train_batteries)

    packs = {
        b: build_e2_pack(
            b,
            vocab,
            condition=condition,
            realization_id=realization_id,
        )
        for b in train_batteries + [test_battery]
    }

    model = build_e4_model(vocab)

    optimizer = torch.optim.AdamW(
        model.parameters(),
        lr=LR,
        weight_decay=WEIGHT_DECAY,
    )

    # Same RNG convention as the canonical experiment.
    rng = np.random.default_rng(seed)

    checkpoint_rows = []
    curve_rows = []

    for epoch in range(1, MAX_EPOCHS + 1):
        order = list(train_batteries)
        rng.shuffle(order)

        model.train()

        epoch_losses = []

        for b in order:
            pack = packs[b]

            cur, pos, neg, prog = make_training_pairs(
                pack["n_ops"],
                NEGATIVES_PER_STATE,
                rng,
            )

            z = model.encode_operations(pack)

            cur_t = torch.as_tensor(
                cur,
                dtype=torch.long,
                device=DEVICE,
            )
            pos_t = torch.as_tensor(
                pos,
                dtype=torch.long,
                device=DEVICE,
            )
            neg_t = torch.as_tensor(
                neg,
                dtype=torch.long,
                device=DEVICE,
            )
            prog_t = torch.as_tensor(
                prog,
                dtype=torch.float32,
                device=DEVICE,
            ).unsqueeze(-1)

            cur_z = z[cur_t]
            pos_z = z[pos_t]

            pos_score = model.score_pairs(
                cur_z,
                pos_z,
                prog_t,
            )

            k = neg_t.shape[1]

            cur_rep = (
                cur_z[:, None, :]
                .expand(-1, k, -1)
                .reshape(-1, cur_z.shape[1])
            )

            neg_z = z[neg_t.reshape(-1)]

            prog_rep = (
                prog_t[:, None, :]
                .expand(-1, k, -1)
                .reshape(-1, 1)
            )

            neg_score = model.score_pairs(
                cur_rep,
                neg_z,
                prog_rep,
            ).reshape(len(cur_z), k)

            loss = F.softplus(
                -(pos_score[:, None] - neg_score)
            ).mean()

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            epoch_losses.append(float(loss.detach().cpu()))

        curve_rows.append({
            "epoch": int(epoch),
            "mean_training_loss": float(np.mean(epoch_losses)),
            "min_battery_loss": float(np.min(epoch_losses)),
            "max_battery_loss": float(np.max(epoch_losses)),
        })

        if epoch in EVAL_EPOCHS:
            metrics, _ = evaluate_neural_model(
                model,
                packs[test_battery],
                train_batteries,
            )

            checkpoint_rows.append({
                "epoch": int(epoch),
                "hits1": float(metrics["hits1"]),
                "hits3": float(metrics["hits3"]),
                "mrr": float(metrics["mrr"]),
                "seen_signature_rate": float(
                    metrics["seen_signature_rate"]
                ),
            })

    return (
        model,
        vocab,
        pd.DataFrame(checkpoint_rows),
        pd.DataFrame(curve_rows),
    )
```

## 16. Smoke test

A short temporary 2-epoch trajectory verifies the four graph conditions before the full run. The global `MAX_EPOCHS` and `EVAL_EPOCHS` values are restored immediately afterward.


```python
# 16. Fast smoke test on one fold

_original_max_epochs = MAX_EPOCHS
_original_eval_epochs = list(EVAL_EPOCHS)

MAX_EPOCHS = 2
EVAL_EPOCHS = [2]

TEST_BATTERY = sorted(chains)[0]
train_batteries = [
    b for b in sorted(chains)
    if b != TEST_BATTERY
]

for condition, rid in [
    ("real", 0),
    ("noedge", 0),
    ("complete", 0),
    ("shuffle", 1),
]:
    model, vocab, checkpoints, curve = train_e4_trajectory(
        train_batteries=train_batteries,
        test_battery=TEST_BATTERY,
        seed=SEEDS[0],
        condition=condition,
        realization_id=rid,
    )

    assert len(checkpoints) == 1
    assert len(curve) == 2

    print(
        condition,
        "H1=",
        float(checkpoints["hits1"].iloc[0]),
        "loss=",
        float(curve["mean_training_loss"].iloc[-1]),
    )

    del model
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

MAX_EPOCHS = _original_max_epochs
EVAL_EPOCHS = _original_eval_epochs

print("Smoke test passed.")
print("Restored EVAL_EPOCHS:", EVAL_EPOCHS)
print("Restored MAX_EPOCHS:", MAX_EPOCHS)
```

## 17. Full E4 run

This is the computationally expensive stage.

There are **240 training trajectories**:

- 30 RealGNN trajectories;
- 30 NoEdgesGNN trajectories;
- 30 CompleteGraphGNN trajectories;
- 150 ShuffledGNN trajectories.

Each trajectory runs to 300 epochs and is evaluated at epochs 30, 100, and 300.

This is equivalent to substantially more computation than E2, so an **L4 or better GPU is strongly recommended**. The notebook writes completed trajectory results immediately, so rerunning the cell after a disconnect skips finished trajectories.


```python
# 17. Full training-budget experiment

OUT = Path("/content/battery_e4_training_budget_lopo")
OUT.mkdir(parents=True, exist_ok=True)

CHECKPOINT_PARTIAL = OUT / "e4_checkpoint_partial.csv"
CURVE_PARTIAL = OUT / "e4_curve_partial.csv"

if CHECKPOINT_PARTIAL.exists():
    checkpoint_results = pd.read_csv(
        CHECKPOINT_PARTIAL
    )
else:
    checkpoint_results = pd.DataFrame()

if CURVE_PARTIAL.exists():
    curve_results = pd.read_csv(
        CURVE_PARTIAL
    )
else:
    curve_results = pd.DataFrame()

battery_ids = sorted(chains)

conditions = [
    ("RealGNN", "real", 0),
    ("NoEdgesGNN", "noedge", 0),
    ("CompleteGraphGNN", "complete", 0),
] + [
    ("ShuffledGNN", "shuffle", sid)
    for sid in SHUFFLE_IDS
]


def trajectory_done(
    df,
    test_battery,
    seed,
    condition,
    realization_id,
):
    if len(df) == 0:
        return False

    sub = df[
        (df["battery_id"] == int(test_battery))
        & (df["seed"] == int(seed))
        & (df["condition"] == str(condition))
        & (df["realization_id"] == int(realization_id))
    ]

    return set(
        sub["epoch"].astype(int).tolist()
    ) == set(EVAL_EPOCHS)


start_time = time.time()
trajectory_counter = 0

for fold_idx, test_battery in enumerate(
    battery_ids,
    start=1,
):
    train_batteries = [
        b for b in battery_ids
        if b != test_battery
    ]

    print(
        f"\nFold {fold_idx}/10 — held out: "
        f"{battery_names[test_battery]}"
    )

    for label, condition, realization_id in conditions:
        for seed in SEEDS:
            trajectory_counter += 1

            if trajectory_done(
                checkpoint_results,
                test_battery,
                seed,
                condition,
                realization_id,
            ):
                print(
                    f"  skip completed {condition} "
                    f"r={realization_id} seed={seed}"
                )
                continue

            print(
                f"  trajectory {trajectory_counter}/{N_TRAJECTORIES}: "
                f"{condition} r={realization_id} seed={seed}"
            )

            model, vocab, checkpoints, curve = (
                train_e4_trajectory(
                    train_batteries=train_batteries,
                    test_battery=test_battery,
                    seed=seed,
                    condition=condition,
                    realization_id=realization_id,
                )
            )

            checkpoints = checkpoints.copy()
            checkpoints["battery_id"] = int(test_battery)
            checkpoints["battery"] = battery_names[test_battery]
            checkpoints["model"] = label
            checkpoints["condition"] = condition
            checkpoints["realization_id"] = int(realization_id)
            checkpoints["seed"] = int(seed)
            checkpoints["parameter_count"] = int(
                count_trainable_parameters(model)
            )

            curve = curve.copy()
            curve["battery_id"] = int(test_battery)
            curve["battery"] = battery_names[test_battery]
            curve["model"] = label
            curve["condition"] = condition
            curve["realization_id"] = int(realization_id)
            curve["seed"] = int(seed)

            # Remove any incomplete duplicate rows for this trajectory,
            # then append the complete result.
            if len(checkpoint_results):
                keep = ~(
                    (checkpoint_results["battery_id"] == int(test_battery))
                    & (checkpoint_results["seed"] == int(seed))
                    & (checkpoint_results["condition"] == str(condition))
                    & (
                        checkpoint_results["realization_id"]
                        == int(realization_id)
                    )
                )
                checkpoint_results = checkpoint_results[keep]

            if len(curve_results):
                keep = ~(
                    (curve_results["battery_id"] == int(test_battery))
                    & (curve_results["seed"] == int(seed))
                    & (curve_results["condition"] == str(condition))
                    & (
                        curve_results["realization_id"]
                        == int(realization_id)
                    )
                )
                curve_results = curve_results[keep]

            checkpoint_results = pd.concat(
                [checkpoint_results, checkpoints],
                ignore_index=True,
            )

            curve_results = pd.concat(
                [curve_results, curve],
                ignore_index=True,
            )

            checkpoint_results.to_csv(
                CHECKPOINT_PARTIAL,
                index=False,
            )

            curve_results.to_csv(
                CURVE_PARTIAL,
                index=False,
            )

            print(
                "    checkpoints:",
                checkpoints[
                    ["epoch", "hits1", "hits3", "mrr"]
                ].round(4).to_dict("records"),
            )

            del model
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

elapsed = (time.time() - start_time) / 3600.0

expected_checkpoint_rows = (
    N_TRAJECTORIES * len(EVAL_EPOCHS)
)

print(
    f"\nCheckpoint rows: {len(checkpoint_results)} "
    f"(expected {expected_checkpoint_rows})"
)
print(
    f"Curve rows: {len(curve_results)} "
    f"(expected {N_TRAJECTORIES * MAX_EPOCHS})"
)
print(f"Elapsed this session: {elapsed:.2f} hours")
```

## 18. Aggregate permutation realizations and neural seeds within held-out product

For `ShuffledGNN`, the five permutation realizations are averaged within each `(product, seed, epoch)` first. Neural seeds are then averaged within product.

This produces one independent held-out-product observation for each condition and training budget.


```python
# 18. Product-level E4 checkpoint summaries

expected_checkpoint_rows = (
    N_TRAJECTORIES * len(EVAL_EPOCHS)
)

assert len(checkpoint_results) == expected_checkpoint_rows, (
    f"E4 incomplete: {len(checkpoint_results)} checkpoint rows "
    f"present; {expected_checkpoint_rows} expected."
)

seed_rows = []

for (
    battery_id,
    battery,
    seed,
    epoch,
), grp in checkpoint_results.groupby(
    [
        "battery_id",
        "battery",
        "seed",
        "epoch",
    ]
):
    for cond in [
        "real",
        "noedge",
        "complete",
    ]:
        g = grp[
            grp["condition"] == cond
        ]
        assert len(g) == 1

        seed_rows.append({
            "battery_id": battery_id,
            "battery": battery,
            "seed": seed,
            "epoch": int(epoch),
            "condition_summary": cond,
            "hits1": float(g["hits1"].iloc[0]),
            "hits3": float(g["hits3"].iloc[0]),
            "mrr": float(g["mrr"].iloc[0]),
            "realization_sd_hits1": 0.0,
        })

    g = grp[
        grp["condition"] == "shuffle"
    ]
    assert len(g) == len(SHUFFLE_IDS)

    seed_rows.append({
        "battery_id": battery_id,
        "battery": battery,
        "seed": seed,
        "epoch": int(epoch),
        "condition_summary": "shuffle_mean",
        "hits1": float(g["hits1"].mean()),
        "hits3": float(g["hits3"].mean()),
        "mrr": float(g["mrr"].mean()),
        "realization_sd_hits1": float(
            g["hits1"].std()
        ),
    })

seed_summary = pd.DataFrame(seed_rows)

fold_mean = (
    seed_summary
    .groupby(
        [
            "battery_id",
            "battery",
            "epoch",
            "condition_summary",
        ],
        as_index=False,
    )
    .agg(
        hits1=("hits1", "mean"),
        hits3=("hits3", "mean"),
        mrr=("mrr", "mean"),
        realization_sd_hits1=(
            "realization_sd_hits1",
            "mean",
        ),
    )
)

aggregate = (
    fold_mean
    .groupby(
        [
            "epoch",
            "condition_summary",
        ],
        as_index=False,
    )
    .agg(
        hits1=("hits1", "mean"),
        hits3=("hits3", "mean"),
        mrr=("mrr", "mean"),
        mean_realization_sd_hits1=(
            "realization_sd_hits1",
            "mean",
        ),
    )
    .sort_values(
        ["epoch", "hits1"],
        ascending=[True, False],
    )
)

display(aggregate)
```

## 19. Exact product-level mechanism contrasts at each training budget

At each checkpoint, the same mechanism contrasts are computed:

- Real − NoEdge
- Permuted − NoEdge
- Complete − NoEdge
- Real − Permuted
- Real − Complete
- Permuted − Complete

Hits@1 is primary for this diagnostic. Holm adjustment is applied separately within each epoch to these six Hits@1 contrasts. Hits@3 and MRR are supporting endpoints.


```python
# 19. Exact paired tests at each training budget

def exact_two_sided_sign_flip(values):
    values = np.asarray(
        values,
        dtype=float,
    )

    observed = abs(
        values.mean()
    )

    total = 2 ** len(values)
    extreme = 0

    for signs in itertools.product(
        (-1.0, 1.0),
        repeat=len(values),
    ):
        stat = abs(
            np.mean(
                values
                * np.asarray(signs)
            )
        )

        if stat >= observed - 1e-15:
            extreme += 1

    return extreme / total


def bootstrap_mean_ci(
    values,
    n_boot=20000,
    seed=20270401,
):
    values = np.asarray(
        values,
        dtype=float,
    )

    rng = np.random.default_rng(seed)

    draws = rng.choice(
        values,
        size=(n_boot, len(values)),
        replace=True,
    ).mean(axis=1)

    return (
        float(
            np.percentile(
                draws,
                2.5,
            )
        ),
        float(
            np.percentile(
                draws,
                97.5,
            )
        ),
    )


def holm_adjust(pvalues):
    pvalues = np.asarray(
        pvalues,
        dtype=float,
    )

    m = len(pvalues)
    order = np.argsort(pvalues)

    out_sorted = np.empty(
        m,
        dtype=float,
    )

    running = 0.0

    for rank, idx in enumerate(order):
        running = max(
            running,
            (m - rank)
            * pvalues[idx],
        )

        out_sorted[rank] = min(
            1.0,
            running,
        )

    out = np.empty(
        m,
        dtype=float,
    )

    for rank, idx in enumerate(order):
        out[idx] = out_sorted[rank]

    return out


contrast_specs = [
    ("Real - NoEdge", "real", "noedge"),
    (
        "Permuted - NoEdge",
        "shuffle_mean",
        "noedge",
    ),
    (
        "Complete - NoEdge",
        "complete",
        "noedge",
    ),
    (
        "Real - Permuted",
        "real",
        "shuffle_mean",
    ),
    (
        "Real - Complete",
        "real",
        "complete",
    ),
    (
        "Permuted - Complete",
        "shuffle_mean",
        "complete",
    ),
]

rows = []

for epoch_idx, epoch in enumerate(
    EVAL_EPOCHS
):
    fm = fold_mean[
        fold_mean["epoch"] == epoch
    ]

    wide = fm.pivot(
        index=[
            "battery_id",
            "battery",
        ],
        columns="condition_summary",
        values=[
            "hits1",
            "hits3",
            "mrr",
        ],
    )

    for metric_idx, metric in enumerate(
        ["hits1", "hits3", "mrr"]
    ):
        for j, (label, left, right) in enumerate(
            contrast_specs
        ):
            d = (
                wide[(metric, left)]
                - wide[(metric, right)]
            ).to_numpy(float)

            ci = bootstrap_mean_ci(
                d,
                seed=(
                    20270401
                    + 1000 * epoch_idx
                    + 100 * metric_idx
                    + j
                ),
            )

            rows.append({
                "epoch": int(epoch),
                "metric": metric,
                "contrast": label,
                "mean_difference": float(
                    d.mean()
                ),
                "ci_low": ci[0],
                "ci_high": ci[1],
                "positive_products": int(
                    (d > 0).sum()
                ),
                "negative_products": int(
                    (d < 0).sum()
                ),
                "zero_products": int(
                    (d == 0).sum()
                ),
                "p_exact": float(
                    exact_two_sided_sign_flip(d)
                ),
            })

contrast_stats = pd.DataFrame(rows)
contrast_stats["p_holm_hits1_within_epoch"] = np.nan

for epoch in EVAL_EPOCHS:
    mask = (
        (contrast_stats["epoch"] == epoch)
        & (
            contrast_stats["metric"]
            == "hits1"
        )
    )

    contrast_stats.loc[
        mask,
        "p_holm_hits1_within_epoch",
    ] = holm_adjust(
        contrast_stats.loc[
            mask,
            "p_exact",
        ].to_numpy(float)
    )

display(
    contrast_stats[
        contrast_stats["metric"]
        == "hits1"
    ]
)
```

## 20. Product-level change with longer training

This section asks whether each condition materially improves or deteriorates as optimization continues. These are descriptive robustness contrasts rather than model-selection tests.

Comparisons:

- 100 − 30 epochs
- 300 − 100 epochs
- 300 − 30 epochs


```python
# 20. Within-condition training-budget changes

budget_pairs = [
    ("100 - 30", 100, 30),
    ("300 - 100", 300, 100),
    ("300 - 30", 300, 30),
]

budget_rows = []

for condition in [
    "real",
    "noedge",
    "complete",
    "shuffle_mean",
]:
    sub = fold_mean[
        fold_mean["condition_summary"]
        == condition
    ]

    wide = sub.pivot(
        index=[
            "battery_id",
            "battery",
        ],
        columns="epoch",
        values=[
            "hits1",
            "hits3",
            "mrr",
        ],
    )

    for metric_idx, metric in enumerate(
        ["hits1", "hits3", "mrr"]
    ):
        for j, (label, later, earlier) in enumerate(
            budget_pairs
        ):
            d = (
                wide[(metric, later)]
                - wide[(metric, earlier)]
            ).to_numpy(float)

            ci = bootstrap_mean_ci(
                d,
                seed=(
                    20270501
                    + 1000
                    * [
                        "real",
                        "noedge",
                        "complete",
                        "shuffle_mean",
                    ].index(condition)
                    + 100 * metric_idx
                    + j
                ),
            )

            budget_rows.append({
                "condition": condition,
                "metric": metric,
                "budget_change": label,
                "mean_difference": float(
                    d.mean()
                ),
                "ci_low": ci[0],
                "ci_high": ci[1],
                "positive_products": int(
                    (d > 0).sum()
                ),
                "negative_products": int(
                    (d < 0).sum()
                ),
                "zero_products": int(
                    (d == 0).sum()
                ),
                "p_exact": float(
                    exact_two_sided_sign_flip(d)
                ),
            })

budget_change_stats = pd.DataFrame(
    budget_rows
)

display(
    budget_change_stats[
        budget_change_stats["metric"]
        == "hits1"
    ]
)
```

## 21. Training-loss convergence summaries

Epoch-level losses are averaged across folds, seeds, and—where applicable—permutation realizations. The loss curves are descriptive because different graph interventions can induce different loss scales and optimization geometry. Their main purpose is to detect continued large improvements or obvious non-convergence after 30 epochs.


```python
# 21. Training-loss summaries

assert len(curve_results) == (
    N_TRAJECTORIES * MAX_EPOCHS
), (
    f"Curve table incomplete: "
    f"{len(curve_results)} rows."
)

curve_summary = (
    curve_results
    .groupby(
        [
            "condition",
            "epoch",
        ],
        as_index=False,
    )
    .agg(
        mean_training_loss=(
            "mean_training_loss",
            "mean",
        ),
        sd_training_loss=(
            "mean_training_loss",
            "std",
        ),
        trajectories=(
            "seed",
            "count",
        ),
    )
)

loss_checkpoints = curve_summary[
    curve_summary["epoch"].isin(
        [1, 10, 30, 50, 100, 200, 300]
    )
].copy()

display(loss_checkpoints)

# Relative loss reduction after the original 30-epoch budget.
loss_wide = curve_summary.pivot(
    index="condition",
    columns="epoch",
    values="mean_training_loss",
)

loss_reduction = pd.DataFrame({
    "condition": loss_wide.index,
    "loss_epoch30": loss_wide[30].values,
    "loss_epoch100": loss_wide[100].values,
    "loss_epoch300": loss_wide[300].values,
})

loss_reduction[
    "relative_reduction_30_to_100"
] = (
    loss_reduction["loss_epoch30"]
    - loss_reduction["loss_epoch100"]
) / loss_reduction["loss_epoch30"]

loss_reduction[
    "relative_reduction_30_to_300"
] = (
    loss_reduction["loss_epoch30"]
    - loss_reduction["loss_epoch300"]
) / loss_reduction["loss_epoch30"]

display(loss_reduction)
```
